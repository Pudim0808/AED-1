```python
vogais = 0
consoantes = 0
frase = input("Digite uma frase: ")
for letra in frase:
  if letra.upper() in "AEIOU":
    vogais += 1
  else:
    consoantes += 1
print(f"{vogais} vogais e {consoantes} consoantes")
```
