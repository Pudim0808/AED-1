```python
def media(nota1, nota2, nota3, aprovacao= 6):
  media = (nota1 + nota2 + nota3) / 3
  if media >= aprovacao:
    resultado = "aprovado"
  else:
    resultado = "reprovado"
  return media, resultado
```

```python
m,r = media(5, 7, 10, 6)
```

```python
print(f"sua media foi {m} e você foi {r}")
```

```python
def vendas(lista_vendas):
  soma = sum(lista_vendas)
  tiket_medio = soma / len(lista_vendas)
  maior_venda = max(lista_vendas)
  menor_venda = min(lista_vendas)
  return soma, tiket_medio, maior_venda
```

```python
v = (456, 324, 540, 12, 89)
s, t, m =vendas(v)
```
