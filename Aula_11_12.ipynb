```python
soma = 0
for A in range(1,6):
  nota = float(input(f"digite a nota {A}: "))
  soma = soma + nota

media = soma / 5
print(f'media: {media:6.2f} ')
```

```python
aprovados = 0
lista_aprovados = []
for i in range(1,6):
  nota = float(input(f"digite a nota {i}: "))
  if nota >= 7:
    aprovados = aprovados + 1
    lista_aprovados.append(nota)
print(f"aprovados: {aprovados}")
for i in lista_aprovados:
  print(i)
```

```python
resposta = input("digite s ou n para continuar: ")
while resposta != "s" and resposta != "n":
  print("resposta invalida")
  resposta = input("digite sim para continuar:")
if resposta == "s":
  print("continua")
else:
  print("para")
```

```python
nota = 0
soma = 0
qtd = 0
while nota != -1:
  nota = float(input("digite a nota: "))
  if nota != -1:
    soma = soma + nota
    qtd = qtd + 1

media = soma / qtd
print(f"media: {media:6.2f}")
```

```python
import random
numero = random.randint(1,10)
palpite = int(input("digite um numero: "))
while palpite != numero:
  if palpite > numero:
    print("menor")
  else:
    print("maior")
  palpite = int(input("digite um numero: "))
print("acertou")
```
