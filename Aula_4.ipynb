```python
def calcular_valor_final(valor_original, desconto_percentual=10):
    desconto = valor_original * (desconto_percentual / 100)
    valor_final = valor_original - desconto
    return valor_final

# Exemplo de uso:
valor_compra = 150.00
valor_com_desconto = calcular_valor_final(valor_compra)
print(f"Valor original: R$ {valor_compra:.2f}")
print(f"Valor com 10% de desconto: R$ {valor_com_desconto:.2f}")
```

# Algoritmo: Preparo de Café Simples (Filtro de Papel)

### **Entradas:**
* Pó de café (2 colheres de sopa)
* Água filtrada (250 ml)
* Filtro de papel
* Suporte para filtro (porta-filtro)
* Garrafa térmica ou caneca
* Fonte de calor (fogão ou chaleira elétrica)

---

### **Passos:**
1. **Ferver a água:** Coloque os 250 ml de água na chaleira e leve ao fogo até começar a borbulhar.
2. **Preparar o filtro:** Encaixe o filtro de papel no suporte e posicione o suporte sobre a garrafa térmica ou caneca.
3. **Escaldar o filtro (Opcional, mas recomendado):** Jogue um pouco de água quente sobre o filtro de papel vazio para retirar o gosto do papel e descarte essa água de enxágue.
4. **Adicionar o café:** Adicione as 2 colheres de sopa de pó de café no filtro de papel e dê uma leve sacudida para nivelar o pó.
5. **Fazer a pré-infusão:** Despeje um pouco de água quente sobre o pó, apenas o suficiente para umedecê-lo, e aguarde 20 segundos.
6. **Coar o café:** Despeje o restante da água quente lentamente em movimentos circulares, do centro para as bordas, e aguarde toda a água passar pelo filtro.

---

### **Saída:**
* Uma caneca ou garrafa com café quente, fresco e pronto para o consumo.

# Análise de Ambiguidades e Refatoração de Algoritmo para Robô

### **1. Identificação das Ambiguidades (4 pontos críticos):**
1. **"Pegue a folha"**: Qual folha? De que tamanho, cor ou material (papel, planta)? Onde ela está localizada antes de ser pega?
2. **"Dobre no meio"**: Como deve ser dobrada? Na horizontal, na vertical ou na diagonal? É necessário alinhar as pontas perfeitamente? Quantas vezes?
3. **"Coloque no lugar certo"**: O que define o "lugar certo"? É uma gaveta, uma caixa de saída, uma mesa específica? Como o robô identifica esse local?
4. **"Entregue para a pessoa"**: Qual pessoa? Como o robô a reconhece (por ID, nome, posição ou reconhecimento facial)? O que fazer se a pessoa não estiver presente?

---

### **2. Algoritmo Refatorado (Mais Preciso):**

**Nome:** Processo de Dobra e Entrega de Documento de Papel

**Entradas:**
* Folha de papel sulfite A4 (localizada na bandeja de entrada "A")
* Coordenadas da gaveta de armazenamento "B" (o "lugar certo")
* Identificador visual (ou facial) da pessoa destinatária "Carlos"

**Passos:**
1. **Localizar e coletar:** Mova o atuador até a bandeja de entrada "A", alinhe a garra sobre a folha de papel sulfite A4 superior e faça a pinça para segurá-la sem rasgar.
2. **Preparar para a dobra:** Posicione a folha de papel sobre uma superfície plana e rígida.
3. **Dobrar com precisão:** Junte a borda lateral inferior à borda lateral superior da folha (dobra horizontal ao meio), alinhe as pontas perfeitamente e pressione o vinco central com o atuador aplicando força constante de 5 Newtons.
4. **Armazenar temporariamente:** Pegue o papel dobrado, mova-se até a gaveta de armazenamento "B" e coloque-o centralizado dentro dela.
5. **Localizar destinatário:** Escaneie o ambiente local usando a câmera integrada até identificar o rosto da pessoa "Carlos" por meio do banco de dados de reconhecimento facial.
6. **Entregar o documento:** Aproxime-se de Carlos, estenda a garra com o papel dobrado na altura de suas mãos e solte a pinça assim que os sensores de pressão detectarem que ele segurou o papel.

**Saída:**
* Folha de papel A4 dobrada ao meio de forma homogênea e entregue nas mãos de Carlos.

# Decomposição de Processo Real e Algoritmo Detalhado

## 1. Processo Escolhido: Lavar a Louça Acumulada

Podemos decompor o processo macro de lavar a louça em 5 partes sequenciais:
1. **Triagem e Organização:** Separar os itens por tipo (copos, talheres, pratos, panelas) e descartar restos de comida no lixo orgânico.
2. **Preparação e Ensaboamento (Parte Escolhida):** Aplicar detergente na esponja e esfregar os utensílios seguindo uma ordem de menor a maior teor de gordura.
3. **Enxágue:** Remover todo o sabão de cada utensílio com água corrente de forma eficiente.
4. **Secagem:** Dispor os itens no escorredor ou secá-los diretamente com um pano de prato seco.
5. **Armazenamento:** Guardar todos os utensílios secos em seus respectivos armários e gavetas.

---

## 2. Algoritmo Detalhado para a Parte 2 (Preparação e Ensaboamento)

**Nome:** Algoritmo de Higienização e Ensaboamento de Utensílios

**Entradas:**
* Utensílios previamente organizados (copos, talheres, pratos, panelas)
* 1 Esponja de lavar louça
* Frasco de detergente líquido
* Torneira com água encanada

**Passos:**
1. **Umedecer e Aplicar:** Abra a torneira brevemente para molhar a esponja e feche-a imediatamente. Aplique aproximadamente uma colher de chá de detergente na parte amarela (macia) da esponja.
2. **Ativar a Espuma:** Aperte a esponja de 3 a 4 vezes com a mão até gerar uma espuma consistente.
3. **Copos e Vidros (Menor Gordura):** Esfregue primeiro os copos e taças usando o lado macio da esponja, garantindo limpar tanto o interior quanto a borda externa.
4. **Talheres e Pratos (Média Gordura):** Esfregue os talheres um a um e, em seguida, passe para os pratos. Use o lado verde (abrasivo) da esponja somente se houver sujeira aderida que não saia com o lado macio.
5. **Panelas e Refratários (Maior Gordura):** Por fim, esfregue as panelas e assadeiras. Caso haja resíduos queimados, use o lado abrasivo com movimentos circulares vigorosos.
6. **Acomodação Temporária:** À medida que cada item for ensaboado, coloque-o cuidadosamente dentro da cuba da pia (ou em uma área de apoio lateral), prontos para o próximo processo (Enxágue).

**Saída:**
* Utensílios completamente ensaboados e prontos para o enxágue, sem resíduos sólidos de alimentos.

# Algoritmo: Organização de uma Festa de Aniversário

### **Entradas:**
* Orçamento disponível (R$)
* Lista de convidados inicial
* Tema ou estilo da festa
* Data e horário pretendidos
* Fornecedores em potencial (comida, bebida, decoração, música)

---

### **Passos:**
1. **Definição de Escopo e Orçamento:** Estipule o valor máximo a ser gasto e defina o número limite de convidados de acordo com esse orçamento.
2. **Escolha e Reserva do Local:** Encontre um espaço adequado ao número de convidados (salão de festas, casa, parque) e confirme a reserva para a data e hora desejadas.
3. **Criação e Envio de Convites:** Elabore um convite físico ou digital detalhando data, hora, local, tema e link para confirmação de presença (RSVP). Envie aos convidados com no mínimo 15 dias de antecedência.
4. **Contratação de Serviços e Compras:** Contrate os fornecedores essenciais (buffet, bolo, decoração, som) ou compre os insumos necessários (comidas, bebidas, descartáveis) se você mesmo for preparar.
5. **Cronograma do Dia do Evento:** Monte uma linha do tempo detalhada com horários de montagem da decoração, chegada dos fornecedores, início da recepção, hora dos parabéns e encerramento.
6. **Recepção e Execução:** No dia da festa, recepcione os convidados, verifique se a comida/bebida está sendo servida adequadamente e garanta que o cronograma planejado seja cumprido.

---

### **Saída:**
* Uma festa realizada dentro do orçamento estipulado, com convidados satisfeitos e registros fotográficos do evento.
