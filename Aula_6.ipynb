### Classificação dos Trechos de Código

* **a) perguntar o nome e exibir uma saudação:** **Sequência** (passos executados linearmente um após o outro).
* **b) verificar se a média é maior ou igual a 5:** **Decisão** (estrutura condicional `if/else`).
* **c) exibir os números de 1 a 10:** **Repetição** (estrutura de laço `for` ou `while`).
* **d) continuar pedindo a senha até a senha correta ser digitada:** **Repetição** (estrutura de laço condicional).

### Algoritmo: Soma de 1 a 10 (com ENQUANTO)

```text
Algoritmo SomarNumeros

Var
    soma, contador : inteiro

Inicio
    soma <- 0
    contador <- 1

    Enquanto contador <= 10 Faca
        soma <- soma + contador
        contador <- contador + 1
    FimEnquanto

    Escreva("A soma dos números de 1 a 10 é: ", soma)
FimAlgoritmo
```

### Correção do Algoritmo com Loop Infinito

No algoritmo original, faltava incrementar a variável de controle `i` dentro do laço `ENQUANTO`. Sem isso, o valor de `i` nunca muda, gerando um loop infinito.

**Pseudocódigo Corrigido:**

```text
ALGORITMO contador
INÍCIO
    i <- 1
  
    ENQUANTO i <= 5 FAÇA
        ESCREVA i
        i <- i + 1  // Incremento necessário para evitar o loop infinito
    FIM_ENQUANTO
FIM
```

### Algoritmo: Menu Interativo de Áreas

```text
Algoritmo MenuAreas

Var
    opcao : inteiro
    raio, areaCirculo : real
    base, altura, areaRetangulo : real

Inicio
    opcao <- -1

    Enquanto opcao <> 0 Faca
        Escreval("\n--- MENU DE OPÇÕES ---")
        Escreval("1. Calcular área do círculo")
        Escreval("2. Calcular área do retângulo")
        Escreval("0. Sair")
        Escreva("Escolha uma opção: ")
        Leia(opcao)

        Escolha opcao
            Caso 1
                Escreva("Digite o raio do círculo: ")
                Leia(raio)
                // Usando pi aproximado como 3.14159
                areaCirculo <- 3.14159 * raio * raio
                Escreval("A área do círculo é: ", areaCirculo)
            Caso 2
                Escreva("Digite a base do retângulo: ")
                Leia(base)
                Escreva("Digite a altura do retângulo: ")
                Leia(altura)
                areaRetangulo <- base * altura
                Escreval("A área do retângulo é: ", areaRetangulo)
            Caso 0
                Escreval("Saindo do programa...")
            OutroCaso
                Escreval("Opção inválida! Tente novamente.")
        FimEscolha
    FimEnquanto
FimAlgoritmo
```
