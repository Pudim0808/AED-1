```python
# 13 / 2 = 6 resto 1
# 6  / 2 = 3 resto 0
# 3  / 2 = 1 resto 1
# 1  / 2 = 0 resto 1
bin(13)
numeros = [13, 42, 7, 100, 255]

for n in numeros:
    print(f'{n:3d}  ->  {bin(n)}  ->  {n:08b}')
```

```python
# Configurações da imagem (Altere manualmente aqui)
largura = int(input("Digite a largura da imagem em pixels: "))
altura = int(input("Digite a altura da imagem em pixels: "))
bytes_por_pixel = 3  # RGB padrão possui 3 bytes por pixel (Red, Green, Blue)

# Cálculos
tamanho_bytes = largura * altura * bytes_por_pixel
tamanho_kb = tamanho_bytes / 1024
tamanho_mb = tamanho_kb / 1024

# Exibição dos resultados formatados
print(f"--- Calculadora de Tamanho de Imagem RGB ---")
print(f"Dimensões: {largura}x{altura} pixels")
print(f"Tamanho em Bytes: {tamanho_bytes:,} bytes")
print(f"Tamanho em KB:    {tamanho_kb:,.2f} KB")
print(f"Tamanho em MB:    {tamanho_mb:,.2f} MB")
```
