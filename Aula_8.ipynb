```python
vendas = float(input("total: R$"))
anos_empresa = float(input("anos: "))
salario_base = 2000.00
comissao = vendas * 0.05
if (vendas > 10000) and (anos_empresa >= 1):
  bonus = 500.00
else:
  bonus = 0.00
novo_salario = salario_base + comissao + bonus
print(f"o salrio inicial é R${salario_base:.2f}")
print(f"o bonus é R${bonus:.2f}")
print(f"o novo salario é R${novo_salario:.2f}")
```

```python
preco_produto = float(input("Digite o preço do produto: R$"))
estado_origem = input("Digite o estado de origem do produto (ex: RJ, SP, MG): ").upper()

# Although the tax rate is 15% in both cases based on the description,
# the conditional structure is kept as requested to 'read qual o estado'
# and for potential future rule changes.
if estado_origem == 'RJ' and preco_produto > 5000:
    imposto_percentual = 0.15
else:
    imposto_percentual = 0.15

valor_imposto = preco_produto * imposto_percentual

print(f"O valor do imposto a ser pago é: R${valor_imposto:.2f}")
```
