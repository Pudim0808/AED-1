```python
def validar_nota(mensagem):
    while True:
        try:
            nota = float(input(mensagem))
            if 0 <= nota <= 10:
                return nota
            else:
                print("Erro: A nota deve estar entre 0 e 10. Tente novamente.")
        except ValueError:
            print("Erro: Entrada inválida. Digite um número decimal.")

def obter_situacao(media):
    if media >= 7.0:
        return "Aprovado"
    elif media >= 5.0:
        return "Recuperação"
    else:
        return "Reprovado"

def cadastrar_alunos():
    alunos = []
    print("--- Cadastro de Alunos ---")
    while True:
        nome = input("Nome do aluno (ou pressione Enter para finalizar): ").strip()
        if not nome:
            break

        # Cadastro de 3 notas por padrão
        notas = []
        for i in range(1, 4):
            nota = validar_nota(f"Digite a nota {i} de {nome}: ")
            notas.append(nota)

        media = sum(notas) / len(notas)
        situacao = obter_situacao(media)

        alunos.append({
            "nome": nome,
            "notas": notas,
            "media": media,
            "situacao": situacao
        })
        print(f"Aluno {nome} cadastrado com sucesso!\n")
    return alunos

def buscar_aluno(alunos):
    print("\n--- Busca de Aluno ---")
    nome_busca = input("Digite o nome do aluno que deseja buscar: ").strip().lower()
    encontrados = [a for a in alunos if nome_busca in a['nome'].lower()]

    if encontrados:
        for aluno in encontrados:
            exibir_relatorio_individual(aluno)
    else:
        print("Nenhum aluno encontrado com esse nome.")

def exibir_relatorio_individual(aluno):
    print(f"\nRelatório de {aluno['nome']}:")
    print(f"  Notas: {aluno['notas']}")
    print(f"  Média Final: {aluno['media']:.2f}")
    print(f"  Situação: {aluno['situacao']}")

def exibir_agrupados_por_situacao(alunos):
    print("\n--- Alunos Agrupados por Situação ---")
    categorias = {"Aprovado": [], "Recuperação": [], "Reprovado": []}
    for aluno in alunos:
        categorias[aluno['situacao']].append(aluno['nome'])

    for situacao, nomes in categorias.items():
        print(f"{situacao} ({len(nomes)}): {', '.join(nomes) if nomes else 'Nenhum'}")

def exibir_ranking(alunos):
    print("\n--- Ranking dos Alunos (Melhores Médias) ---")
    ranking = sorted(alunos, key=lambda x: x['media'], reverse=True)
    for i, aluno in enumerate(ranking, 1):
        print(f"{i}º - {aluno['nome']}: Média {aluno['media']:.2f} ({aluno['situacao']})")

def exibir_estatisticas_gerais(alunos):
    if not alunos:
        print("Sem dados suficientes para estatísticas.")
        return

    medias = [aluno['media'] for aluno in alunos]
    media_geral = sum(medias) / len(alunos)
    maior_media = max(medias)
    menor_media = min(medias)

    print("\n--- Estatísticas Gerais da Turma ---")
    print(f"Total de alunos cadastrados: {len(alunos)}")
    print(f"Média geral da turma: {media_geral:.2f}")
    print(f"Maior média alcançada: {maior_media:.2f}")
    print(f"Menor média alcançada: {menor_media:.2f}")

def menu():
    alunos = []
    # Inicializa com alguns dados de teste para facilitar a execução rápida
    alunos.append({"nome": "Alice", "notas": [8.0, 9.0, 7.5], "media": 8.17, "situacao": "Aprovado"})
    alunos.append({"nome": "Bruno", "notas": [5.0, 6.0, 4.5], "media": 5.17, "situacao": "Recuperação"})
    alunos.append({"nome": "Carlos", "notas": [3.0, 4.0, 2.5], "media": 3.17, "situacao": "Reprovado"})

    while True:
        print("\n================ PROJETO 3 ================")
        print("1. Cadastrar Alunos")
        print("2. Buscar Aluno pelo Nome")
        print("3. Exibir Alunos Agrupados por Situação")
        print("4. Exibir Ranking (Melhores Médias)")
        print("5. Exibir Estatísticas Gerais")
        print("6. Sair")
        print("===========================================")

        opcao = input("Escolha uma opção (1-6): ").strip()

        if opcao == '1':
            alunos.extend(cadastrar_alunos())
        elif opcao == '2':
            buscar_aluno(alunos)
        elif opcao == '3':
            exibir_agrupados_por_situacao(alunos)
        elif opcao == '4':
            exibir_ranking(alunos)
        elif opcao == '5':
            exibir_estatisticas_gerais(alunos)
        elif opcao == '6':
            print("Encerrando o sistema. Até mais!")
            break
        else:
            print("Opção inválida! Tente novamente.")

# Executar o sistema interactivo
menu()
```
