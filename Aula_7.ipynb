```python
faturamento_total = float(input("Digite o faturamento total da loja: "))
quantidade_pedidos = float(input("Digite a quantidade de pedidos: "))

if quantidade_pedidos > 0:
    ticket_medio = faturamento_total / quantidade_pedidos
    print(f"O ticket médio da loja é: R$ {ticket_medio:.2f}")
else:
    print("A quantidade de pedidos deve ser maior que zero para calcular o ticket médio.")
```
