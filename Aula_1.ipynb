### Explicação: A Função `print()`

A função `print()` em Python é utilizada para enviar e exibir dados na tela. Ela pode receber diversos argumentos de entrada separados por vírgula e os imprime como texto.

**Pontos Importantes:**
* **Textos (Strings):** Devem sempre estar entre aspas simples (`'Olá'`) ou aspas duplas (`"Olá"`).
* **Números e Expressões:** Podem ser passados diretamente, sem aspas (ex: `print(2025)` ou `print(2 + 2)`).
* **Variáveis:** Você pode passar o nome de uma variável para exibir o valor que ela armazena.

```python
# Exemplo 1: Imprimindo uma mensagem de texto simples
print('Olá, este é um texto simples!')

# Exemplo 2: Imprimindo números e resultados de operações matemáticas
print('Resultado de 5 + 5 =', 5 + 5)

# Exemplo 3: Usando variáveis para compor a mensagem
nome = 'Camilo'
curso = 'CDN'
print(f'O aluno {nome} está matriculado no curso {curso}.')
```
