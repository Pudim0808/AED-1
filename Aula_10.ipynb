```python
produto = input("Produto: ")
preco = float(input("Preço: "))
quantidade = int(input("Quantidade: "))
total = preco * quantidade

print("\nTipos das variáveis ao final da execução:")
print(f"produto: {type(produto)}")
print(f"preco: {type(preco)}")
print(f"quantidade: {type(quantidade)}")
print(f"total: {type(total)}")
```

```python
# Leitura da temperatura
temperatura = float(input("Digite a temperatura em °C: "))

# Classificação da temperatura
if temperatura < 15:
    classificacao = "frio"
elif temperatura < 25:
    classificacao = "agradável"
else:
    classificacao = "quente"

# Exibição do resultado
print(f"A temperatura de {temperatura}°C é considerada: {classificacao}.")
```

```python
# Código corrigido:
# Convertemos a entrada para float para permitir a comparação numérica e aceitar valores decimais
valor = float(input("Valor: "))

if valor >= 100:
    print("Compra alta")
else:
    print("Compra comum")
```

```python
def classificar_compra(valor):
    if valor < 100.0:
        return "pequena"
    elif valor < 500.0:
        return "média"
    else:
        return "grande"

# Valores para teste solicitados
valores_teste = [100.00, 500.00]

for valor in valores_teste:
    classificacao = classificar_compra(valor)
    print(f"Uma compra de R$ {valor:,.2f} é classificada como: {classificacao.upper()}")
```
