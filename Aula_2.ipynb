### Entendendo a diferença entre Bit e Byte

Imagine que estamos em um restaurante e queremos representar as informações de forma muito simples:

* **Bit (a menor unidade):** Pense no **Bit** como um **interruptor de luz** na parede. Ele só tem dois estados possíveis: ou está desligado (representando o número `0`) ou está ligado (representando o número `1`). Sozinho, um interruptor não consegue guardar muita informação — ele apenas diz "sim" ou "não".

* **Byte (o grupo de bits):** Agora, imagine que juntamos **8 interruptores lado a lado**. Esse conjunto de 8 interruptores é um **Byte**. Combinando quais interruptores estão ligados ou desligados, conseguimos formar códigos bem mais complexos. É exatamente assim que o computador escreve uma única letra (como a letra 'A'), um número ou um símbolo.

**Em resumo:**
* O **Bit** é uma única lâmpada (acesa ou apagada).
* O **Byte** é um painel com 8 lâmpadas que, juntas, formam uma letra ou um caractere legível.
