### Associação de Ações aos Símbolos de Fluxograma

* **a) Ler a idade do usuário:** **Entrada manual / Entrada de dados** (Paralelogramo)
* **b) Calcular media = (n1 + n2) / 2:** **Processo** (Retângulo)
* **c) Verificar se idade >= 18:** **Decisão** (Losango)
* **d) Marcar o início do algoritmo:** **Terminal** (Elipse / Retângulo Arredondado)
* **e) Exibir "Cadastro aprovado":** **Saída de dados / Exibição** (Paralelogramo ou símbolo de Documento/Display)

### Fluxograma em Formato de Texto (Algoritmo de Desconto)

**[INÍCIO]** (Terminal)
  * **PASSO 1:** Ler o valor da compra e armazenar em `valor_compra` (Entrada de dados)
  * **PASSO 2:** **VERIFICAÇÃO / DECISÃO:**
    * **SE** `valor_compra > 200.00`:
      * Calcular `desconto = valor_compra * 0.10` (Processo)
      * Calcular `valor_final = valor_compra - desconto` (Processo)
    * **SENÃO**:
      * Definir `valor_final = valor_compra` (Processo)
  * **PASSO 3:** Exibir o valor final calculado em `valor_final` (Saída de dados)
**[FIM]** (Terminal)
